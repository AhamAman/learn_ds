1. Simple Random Sampling: 100 records randomly

```python
sample_data = data.sample(n = 100, random_state=44)
sample_data
```

2. Stratified Random Sampling

```python
stratified_sample_data = data.groupby('Region').apply(lambda x : x.sample(min(len(x),25), random_state=44))
stratified_sample_data.to_csv('stratified_sample_data.csv')
```

Cluster Sampling: Manufacturing Data by Batch Number

```python
manufacturing_data = pd.read_csv('/content/drive/MyDrive/ChaiCode Full Stack Data Science Cohort/Phase 1 - Statistics for Data Science/DescriptiveStats_18June_7PM/Manufacturing_Data.csv')
manufacturing_data.head()
```

```python
clusters = manufacturing_data['BatchNumber'].unique()
clusters
```

```python
selected_clusters = np.random.choice(clusters, size = 5, replace = False)
selected_clusters
```

```python
clustered_sample = manufacturing_data[manufacturing_data['BatchNumber'].isin(selected_clusters)]
print(clustered_sample)
```

Systematic Sampling

```python
systematic_sample = data.iloc[::10, :]
systematic_sample
```
