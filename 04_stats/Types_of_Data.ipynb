Data Creation of the nominal column named 'colors'

```python
import pandas as pd
import numpy as np

np.random.seed(0)
num_samples = 1000
colors = ['Red', 'Blue', 'Green', 'Yellow', 'Crimson', 'Purple']
nominal_data = np.random.choice(colors, num_samples)
nominal_data
```

```python
dataframe = pd.DataFrame(nominal_data, columns = ['Colors'])
dataframe
```

Data creation of the ordinal column named 'satisfaction_levels'

satisfaction_levels = ['Low', 'Medium', 'High']

```python
satisfaction_level = ['Low', 'Medium', 'High']
ordinal_data = np.random.choice(satisfaction_level, size=num_samples)
dataframe['satisfaction_levels'] = ordinal_data
dataframe.head()
```

Data creation of the interval data under Temperature column

```python
np.random.seed(0)
interval_data = np.random.randint(0, 100, num_samples)
dataframe['temperature'] = interval_data
dataframe
```

Data creation of the range data under the income column

```python
np.random.seed(0)
ratio_data = np.random.randint(10000,10000000,num_samples)
dataframe['income'] = ratio_data
dataframe
```

Save the file in the csv

```python
dataframe.to_csv('types_of_data.csv')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

## Plot Ratio Data
plt.hist(dataframe['income'], bins=20, color='green', alpha=0.7, edgecolor='black')
plt.title('Income Distribution')
plt.xlabel('Income (Rupees)')
plt.ylabel('Frequency')
plt.savefig('income_distribution.png')
plt.show()
```
