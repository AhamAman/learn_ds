Arithmetic Operations

```python
import numpy as np
a = np.array([[1, 2], [3, 4]])
b = np.array([[10, 20], [30, 40]])

print(a)

print(b)


print(a + b)  # Addition



print(a - b)  # Subtraction



print(a * b)  # Element-wise multiplication




print(a / b)  # Element-wise division



print(a ** 2) # Power

print(b // a)  # Floor division
```

Sorting

```python
arr_1D = np.array([2, 1, 3, 5, 4, 10, 7, 6])
sorted_arr_1D = np.sort(arr_1D)
print(sorted_arr_1D)
```

```python
arr_2D = np.array([[3, 2, 1],
                   [6, 5, 4]])
sorted_arr_2D = np.sort(arr_2D)
print(sorted_arr_2D)
```

```python
arr_2D = np.array([[3, 2, 1],
                   [6, 5, 4]])
sorted_arr_2D = np.sort(arr_2D, axis=1)
print(sorted_arr_2D)
```

axis=0 runs vertically down the columns

axis=1 runs horizontally across the rows

```python
arr_2D = np.array([[3, 2, 1],
                   [6, 5, 4]])
sorted_arr_2D = np.sort(arr_2D, axis=0)
print(sorted_arr_2D)
```

Ranking in Recommendation Systems

```python
similarity_scores = np.array([0.23, 0.85, 0.11, 0.65, 0.91])
sorted_indices = np.argsort(similarity_scores)
## print(sorted_indices)
top_3_items = sorted_indices[-3:][::-1]
print(top_3_items)
```

Concatenation Arrays

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

result = np.concatenate((arr1, arr2))
print(result)
```

```python
arr1 = np.array([[1, 2, 3], [3, 4, 5]])
arr2 = np.array([[4, 5, 6], [5, 6, 7]])

result = np.concatenate((arr1, arr2), axis=0)
print(result)
```

```python
arr1 = np.array([[1, 2, 3], [3, 4, 5]])
arr2 = np.array([[4, 5, 6], [5, 6, 7]])

result = np.concatenate((arr1, arr2), axis=1)
print(result)
```

Statistical Operations: Mean, Median and Mode, Standard Deviation, Min, Max, Sum

np.median => Median

np.std => Standard Deviation

np.min => Minimum Value

np.max => Maximum Value

np.sum => Sum

```python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])

mean_all = np.mean(arr)
print(mean_all)
```

```python
mean_axis_0 = np.mean(arr, axis=0) ## axis=0 indicates column-wise
print(mean_axis_0)
```

```python
mean_axis_1 = np.mean(arr, axis=1) ## axis=1 indicates row-wise
print(mean_axis_1)
```
